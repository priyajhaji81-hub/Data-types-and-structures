## Part A: Theory Questions

### 1. What are data structures, and why are they important?
Data structures are ways of organizing and storing data in a computer so that it can be accessed and modified efficiently. They help make programs faster and easier to manage.
*Examples*: Lists, tuples, dictionaries, and sets.

### 2. Explain the difference between mutable and immutable data types with examples.
* **Mutable**: Data that can be changed after creation. Examples: lists, dictionaries, and sets.
* **Immutable**: Data that cannot be changed after creation. Examples: strings, tuples, and integers.

```python
# Example of Mutable vs Immutable
my_list = [1, 2, 3]
my_list[0] = 10
print("Mutable (List):", my_list)  # [10, 2, 3]

my_string = "Hello"
try:
    my_string[0] = "Y"  # This will raise a TypeError
except TypeError as e:
    print("Immutable (String Error):", e)
```

### 3. What are the main differences between lists and tuples in Python?

| Feature | List | Tuple |
| :--- | :--- | :--- |
| **Mutability** | Mutable | Immutable |
| **Syntax** | Created using `[]` | Created using `()` |
| **Modification** | Elements can be changed | Elements cannot be reassigned |
| **Example** | `[1, 2, 3]` | `(1, 2, 3)` |

### 4. Describe how dictionaries store data.
Dictionaries store data in key-value pairs. Each key is unique and is used to access its associated value.

```python
student = {"name": "Alice", "age": 20}
print(student["name"])  # Alice
```

### 5. Why might you use a set instead of a list in Python?
A set is useful when you want to store only unique values and perform operations such as union and intersection. Sets also provide efficient membership checking on average.

```python
numbers = {1, 2, 2, 3, 3}
print(numbers)  # {1, 2, 3}
```

### 6. What is a string in Python, and how is it different from a list?
A string is a sequence of characters enclosed in single or double quotes. A list is a collection that can contain elements of different data types. Strings are immutable, whereas lists are mutable.

### 7. How do tuples ensure data integrity in Python?
Tuples help protect the structure of data because their elements cannot be reassigned, added, or removed after creation. They are useful for storing fixed collections of values.
*Note*: A tuple containing a mutable object, such as a list, does not make that inner object immutable.

### 8. What is a hash table, and how does it relate to dictionaries in Python?
A hash table is a data structure that uses a hash function to organize keys for efficient lookup. Python dictionaries use hash-table-based implementation to retrieve values using their keys.

### 9. Can lists contain different data types in Python?
Yes. Python lists can contain elements of different data types, including integers, strings, floats, and even other lists.

```python
my_list = [10, "Hello", 3.5, True]
print(my_list)
```

### 10. Explain why strings are immutable in Python.
Strings are immutable because their characters cannot be changed after the string is created. This helps make strings predictable and allows them to be used safely as dictionary keys.

### 11. What advantages do dictionaries offer over lists for certain tasks?
Dictionaries allow data to be accessed using meaningful keys rather than numeric positions. They are useful for storing related information and finding values efficiently.

```python
employee = {"name": "John", "salary": 30000}
print(employee["salary"])
```

### 12. How do sets handle duplicate values in Python?
Sets automatically keep only one occurrence of each element. Duplicate values are removed when the set is created.

### 13. Describe a scenario where using a tuple would be preferable over a list.
A tuple is preferable when storing fixed data that should not be changed, such as RGB color values, geographic coordinates, or days of the week.

### 14. How does the `in` keyword work differently for lists and dictionaries?
* In a list, `in` checks whether a value exists among the elements.
* In a dictionary, `in` checks whether a key exists, not whether a value exists.

```python
numbers = [1, 2, 3]
print(2 in numbers)  # True

student = {"name": "Alice", "age": 20}
print("name" in student)  # True
print("Alice" in student)  # False
```

### 15. Can you modify the elements of a tuple? Explain why or why not.
No. You cannot reassign, add, or remove elements from a tuple because tuples are immutable.

### 16. What is a nested dictionary, and give an example of its use case.
A nested dictionary is a dictionary that contains another dictionary as a value. It is useful for organizing related information about multiple employees or students.

```python
students = {
    "student1": {"name": "Alice", "age": 20},
    "student2": {"name": "Bob", "age": 22}
}
print(students["student1"]["name"])  # Alice
```

### 17. Describe the time complexity of accessing elements in a dictionary.
Dictionary lookup takes O(1) average time, meaning it usually takes constant time to find a value using its key. In the worst case, lookup can take O(n), depending on hash collisions and implementation details.

### 18. In what situations are lists preferred over dictionaries?
Lists are preferred when:
1. Data needs to be stored in a specific sequence.
2. Duplicate values are allowed.
3. You need to access elements by their position.
4. You need to iterate through a collection of items.

### 19. Why are dictionaries considered unordered, and how does that affect data retrieval?
Traditionally, dictionaries were described as unordered because they were not intended for position-based retrieval. However, Python 3.7 and later guarantee insertion order. Values are retrieved by keys rather than positions, making key-based lookup efficient.

### 20. Explain the difference between a list and a dictionary in terms of data retrieval.
A list retrieves elements using numeric indexes, whereas a dictionary retrieves values using keys.

```python
my_list = ["Apple", "Banana", "Mango"]
print(my_list[1])  # Banana

my_dict = {"fruit": "Apple", "color": "Red"}
print(my_dict["color"])  # Red
```
